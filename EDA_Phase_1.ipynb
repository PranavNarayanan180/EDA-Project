```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Loading the Dataset

```python
df = pd.read_csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/1000%20Sales%20Records.csv")
```

```python
print("\nFirst 5 rows of the dataset:")
display(df.head())
```

## 2. Applying basic statistical analysis

```python
print("Dataset Info:")
print(df.info())
print("\nFirst 5 rows of the dataset:")
display(df.head())
print("\nSummary Statistics for Numerical Columns:")
display(df.describe())
print("\nSummary Statistics for Categorical Columns:")
display(df.describe(include=['O']))
```

## 3. Handling missing data

```python
print("Missing values before handling:")
print(df.isnull().sum())

for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        median_val = df[col].median()

        if pd.isna(median_val):
            median_val = 0
        df[col] = df[col].fillna(median_val)
    else:
        mode_val = df[col].mode()
        if len(mode_val) > 0:
            df[col] = df[col].fillna(mode_val[0])
        else:
            df[col] = df[col].fillna("Unknown")

print("\nMissing values after universal handling:")
print(df.isnull().sum())
print("\n")
```

## 4. Data cleaning

```python
# Strip leading/trailing whitespaces
df.columns = df.columns.str.strip()

# Strip whitespaces from string
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

# Drop duplicate rows
initial_rows = len(df)
df = df.drop_duplicates()
print(f"Dropped {initial_rows - len(df)} duplicate rows.\n")
```

## 5. Data transformation

```python
# Convert Order Date and Ship Date to datetime objects
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Create a new feature: Processing/Shipping Time in days
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

# Extract the Year and Month from Order Date for temporal analysis
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month

display(df.head())
```

```python
# 1. Decimal Scaling
max_abs_val = df['Units Sold'].abs().max()
if max_abs_val > 0:
    j = np.ceil(np.log10(max_abs_val))
else:
    j = 1
df['Units Sold_Decimal'] = df['Units Sold'] / (10**j)

# 2. Min-Max Scaling (0 to 1)
min_val = df['Units Sold'].min()
max_val = df['Units Sold'].max()
df['Units Sold_MinMax'] = (df['Units Sold'] - min_val) / (max_val - min_val)

# 3. Z-Score Scaling
mean_val = df['Units Sold'].mean()
std_val = df['Units Sold'].std()
df['Units Sold_ZScore'] = (df['Units Sold'] - mean_val) / std_val

display(df.head())
```

## K-Means

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Load the dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/1000%20Sales%20Records.csv"
df = pd.read_csv(url)

# 2. Remove null values
df_clean = df.dropna().copy()
```

```python
num_cols = df_clean.select_dtypes(include=[np.number]).columns
X = df_clean[num_cols]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

wcss = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(9, 5))
plt.plot(K_range, wcss, marker='o', linestyle='--', color='b')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('WCSS')
plt.xticks(K_range)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
optimal_k = 2

final_kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df_clean['Cluster'] = final_kmeans.fit_predict(X_scaled)

print(f"Cluster sizes for k={optimal_k}:")
print(df_clean['Cluster'].value_counts())

plt.figure(figsize=(9, 6))
sns.scatterplot(
    x='Total Revenue',
    y='Total Profit',
    hue='Cluster',
    palette='viridis',
    data=df_clean,
    alpha=0.7
)
plt.title('K-Means Clusters (Total Revenue vs Total Profit)')
plt.show()
```

## Hierarchical clustering

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

# 1. Load the dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/1000%20Sales%20Records.csv"
df = pd.read_csv(url)

# 2. Remove null values
df_clean = df.dropna().copy()
```

```python
if 'Order ID' in df_clean.columns:
    df_clean = df_clean.drop('Order ID', axis=1)

# 4. Select numerical columns
num_cols = df_clean.select_dtypes(include=[np.number]).columns
X = df_clean[num_cols]

# 5. Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"Data scaled. Shape of features used for clustering: {X_scaled.shape}")
```

```python
plt.figure(figsize=(12, 7))
dendrogram = sch.dendrogram(sch.linkage(X_scaled, method='ward'))

plt.title('Dendrogram for Hierarchical Clustering')
plt.xlabel('Sales Records (Indices)')
plt.ylabel('Euclidean Distances')

plt.axhline(y=15, color='r', linestyle='--')
plt.show()
```

```python
optimal_k_hc = 2
hc = AgglomerativeClustering(n_clusters=optimal_k_hc, metric='euclidean', linkage='ward')
df_clean['HC_Cluster'] = hc.fit_predict(X_scaled)


print("Hierarchical Cluster sizes:")
print(df_clean['HC_Cluster'].value_counts())
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    x='Total Revenue',
    y='Total Profit',
    hue='HC_Cluster',
    palette='Set1',
    data=df_clean,
    alpha=0.8
)

plt.title('Hierarchical Clustering (Total Revenue vs Total Profit)')
plt.xlabel('Total Revenue')
plt.ylabel('Total Profit')
plt.legend(title='Cluster')
plt.show()
```
