```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Loading the Dataset

```python
df = pd.read_csv("https://raw.githubusercontent.com/salemprakash/EDA/main/Data/1000%20Sales%20Records.csv")
```

```python
print("\nFirst 5 rows of the dataset:")
display(df.head())
```

## 2. Applying basic statistical analysis

```python
print("Dataset Info:")
print(df.info())
print("\nFirst 5 rows of the dataset:")
display(df.head())
print("\nSummary Statistics for Numerical Columns:")
display(df.describe())
print("\nSummary Statistics for Categorical Columns:")
display(df.describe(include=['O']))
```

## 3. Handling missing data

```python
print("Missing values before handling:")
print(df.isnull().sum())

for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        median_val = df[col].median()

        if pd.isna(median_val):
            median_val = 0
        df[col] = df[col].fillna(median_val)
    else:
        mode_val = df[col].mode()
        if len(mode_val) > 0:
            df[col] = df[col].fillna(mode_val[0])
        else:
            df[col] = df[col].fillna("Unknown")

print("\nMissing values after universal handling:")
print(df.isnull().sum())
print("\n")
```

## 4. Data cleaning

```python
# Strip leading/trailing whitespaces
df.columns = df.columns.str.strip()

# Strip whitespaces from string
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].str.strip()

# Drop duplicate rows
initial_rows = len(df)
df = df.drop_duplicates()
print(f"Dropped {initial_rows - len(df)} duplicate rows.\n")
```

## 5. Data transformation

```python
# Convert Order Date and Ship Date to datetime objects
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Create a new feature: Processing/Shipping Time in days
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

# Extract the Year and Month from Order Date for temporal analysis
df['Order Year'] = df['Order Date'].dt.year
df['Order Month'] = df['Order Date'].dt.month

display(df.head())
```

## 6. Univariate analysis with a minimum of three visualizations

```python
# 1. Histogram of Total Profit
plt.figure(figsize=(10, 5))
sns.histplot(df['Total Profit'], bins=30, kde=True, color='blue')
plt.title('1. Distribution of Total Profit')
plt.xlabel('Total Profit')
plt.ylabel('Frequency')
plt.show()

# 2. Countplot of Orders by Region
plt.figure(figsize=(12, 6))
sns.countplot(y=df['Region'], order=df['Region'].value_counts().index, palette='viridis')
plt.title('2. Number of Orders per Region')
plt.xlabel('Count')
plt.ylabel('Region')
plt.show()

# 3. Boxplot of Unit Price
plt.figure(figsize=(10, 5))
sns.boxplot(x=df['Unit Price'], color='skyblue')
plt.title('3. Boxplot of Unit Price')
plt.xlabel('Unit Price')
plt.show()
```

## 7. Bivariate analysis with a minimum of three visualizations

```python
# 1. Scatter plot: Total Revenue vs Total Profit
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Total Revenue', y='Total Profit', data=df, alpha=0.6, color='green')
plt.title('1. Total Revenue vs Total Profit')
plt.xlabel('Total Revenue')
plt.ylabel('Total Profit')
plt.show()

# 2. Bar plot: Average Profit by Item Type
plt.figure(figsize=(12, 6))
sns.barplot(x='Total Profit', y='Item Type', data=df, errorbar=None, palette='magma', order=df.groupby('Item Type')['Total Profit'].mean().sort_values(ascending=False).index)
plt.title('2. Average Total Profit by Item Type')
plt.xlabel('Average Total Profit')
plt.ylabel('Item Type')
plt.show()

# 3. Boxplot: Total Profit across Order Priority
plt.figure(figsize=(10, 6))
sns.boxplot(x='Order Priority', y='Total Profit', data=df, palette='Set2')
plt.title('3. Total Profit Distribution by Order Priority')
plt.xlabel('Order Priority')
plt.ylabel('Total Profit')
plt.show()
```

## 8. Multivariate analysis with a minimum of three visualizations

```python
# 1. Correlation Heatmap for numerical variables
plt.figure(figsize=(12, 8))
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('1. Correlation Heatmap of Numeric Features')
plt.show()

# 2. Scatter plot: Total Revenue vs Total Profit, differentiated by Sales Channel (Hue)
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Total Revenue', y='Total Profit', hue='Sales Channel', data=df, alpha=0.7, palette='deep')
plt.title('2. Total Revenue vs Total Profit by Sales Channel')
plt.xlabel('Total Revenue')
plt.ylabel('Total Profit')
plt.legend(title='Sales Channel')
plt.show()

# 3. Pairplot: Pairwise relationships for selected key metrics, colored by Order Priority
selected_columns = ['Total Revenue', 'Total Cost', 'Total Profit', 'Shipping Days', 'Order Priority']
sns.pairplot(df[selected_columns], hue='Order Priority', palette='Set1', corner=True)
plt.suptitle('3. Pairplot of Key Financial Metrics by Order Priority', y=1.02)
plt.show()
```
